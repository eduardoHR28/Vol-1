```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df = pd.read_csv('/content/ecommerce_preparados.csv')

print(df.head().to_string())

# Histograma (registra a distribuição de frequência de dados contínuos)
plt.figure(figsize = (10,6))
plt.title('Histograma de Descontos')
plt.xlabel('Descontos')
plt.ylabel('Quantidade')
plt.hist(df['Desconto'])
plt.show()

# Dispersão (registra a dispersão de dados entre duas variáveis)
plt.figure(figsize = (10,6))
plt.title('Dispersão de Nota com o Desconto')
plt.xlabel('Nota')
plt.ylabel('Desconto')
plt.scatter(df['Nota'], df['Desconto'])
plt.show()

# Mapa de calor
plt.figure(figsize = (10,6))
plt.title('Mapa de Calor da Correlação entre as Propriedades')
df_corr = df[['Nota','Desconto','Preço','Marca_Cod','Material_Cod','Temporada_Cod','Qtd_Vendidos_Cod']].corr()
sns.heatmap(df_corr, annot=True, fmt='.2f')
plt.show()

# Gráfico de Barra
plt.xlabel('Temporada_Cod')
plt.ylabel('Contagem')
plt.legend(title='Contagem de Peças por Temporada')
sns.countplot(x='Temporada_Cod', hue='Temporada' ,data=df, palette='pastel')
plt.show()

# Gráfico de Pizza
x = df['Temporada'].value_counts().index
y = df['Temporada_Cod'].value_counts().values
plt.figure(figsize=(10,6))
plt.pie(y, labels=x, autopct='%.1f%%', startangle=90)   # autopct diz da porcentager ser até .1%,
plt.title('Gráfico de Pizza por Temporada')          # e o startangle em qual angulo começa
plt.figure(figsize=(10,6))
plt.show()

# Gráfico de Densidade
plt.figure(figsize=(10,10))
sns.kdeplot(df['Preço'], fill=True, color='red')
plt.title('Densidade de Preços')
plt.xlabel('Preço')
plt.show()

# Gráfico  de Regressão
plt.figure(figsize=(10,10))
sns.regplot(x='Nota', y='Desconto', data=df, color='green', scatter_kws={'alpha': 0.3, 'color':'red'})
plt.title('Gráfico de Regressão Linear Nota x Desconto')
plt.xlabel('Nota')
plt.ylabel('Desconto')
plt.show()

```
